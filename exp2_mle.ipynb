## Objective
To understand and implement Maximum Likelihood Estimation (MLE) for estimating parameters of a Gaussian distribution.

## Theory
MLE is a method of estimating the parameters of a statistical model. Given observed data, MLE finds the parameter values that **maximize the likelihood** of observing that data.

For a Gaussian distribution:
- **MLE of mean (μ̂)** = sample mean = (1/n) Σxᵢ
- **MLE of std (σ̂)** = sample standard deviation = √[(1/n) Σ(xᵢ - μ̂)²]

MLE is the foundation of training most generative models (GANs, VAEs, Flow models).

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

sns.set(style='whitegrid', font_scale=1.2)
np.random.seed(42)

# --- True Parameters ---
true_mean = 5.0
true_std  = 2.0
n_samples = 1000

data = np.random.normal(true_mean, true_std, n_samples)

# --- MLE Estimates ---
mle_mean = np.mean(data)
mle_std  = np.std(data)   # MLE std (divides by N)

print(f"True  -> Mean: {true_mean:.4f} | Std: {true_std:.4f}")
print(f"MLE   -> Mean: {mle_mean:.4f} | Std: {mle_std:.4f}")
print(f"Error -> Mean: {abs(true_mean - mle_mean):.4f} | Std: {abs(true_std - mle_std):.4f}")

# --- Plot ---
x = np.linspace(data.min() - 1, data.max() + 1, 300)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Maximum Likelihood Estimation', fontsize=15, fontweight='bold')

# Plot 1: Histogram + MLE fitted curve
sns.histplot(data, kde=False, stat='density', ax=axes[0], color='lightblue', label='Data')
axes[0].plot(x, norm.pdf(x, true_mean, true_std), 'b-', lw=2, label=f'True N({true_mean},{true_std})')
axes[0].plot(x, norm.pdf(x, mle_mean, mle_std),  'r--', lw=2, label=f'MLE N({mle_mean:.2f},{mle_std:.2f})')
axes[0].axvline(mle_mean, color='red', linestyle=':', label='MLE Mean')
axes[0].axvline(true_mean, color='blue', linestyle=':', label='True Mean')
axes[0].set_title('Data with True vs MLE Distribution')
axes[0].legend(fontsize=9)

# Plot 2: Log-Likelihood vs mean (std fixed)
means = np.linspace(true_mean - 2, true_mean + 2, 100)
log_likelihoods = [np.sum(norm.logpdf(data, m, true_std)) for m in means]
axes[1].plot(means, log_likelihoods, color='purple', lw=2)
axes[1].axvline(mle_mean, color='red', linestyle='--', label=f'MLE Mean={mle_mean:.2f}')
axes[1].set_title('Log-Likelihood vs Mean (σ fixed)')
axes[1].set_xlabel('Mean (μ)')
axes[1].set_ylabel('Log-Likelihood')
axes[1].legend()

plt.tight_layout()
plt.savefig('exp2_mle.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Result
- The MLE estimates are very close to the true parameters.
- The log-likelihood curve peaks exactly at the MLE mean, confirming the estimate.

## Conclusion
MLE provides a principled way to estimate distribution parameters from data. With larger samples, MLE estimates converge to the true values (consistency property). This is the backbone of training all probabilistic generative models.
