## Objective
To understand and visualize different types of probability distributions used in generative AI models.

## Theory
Probability distributions describe how data is spread across possible values. In generative AI:
- **Normal (Gaussian) Distribution**: Used in VAEs as the latent space prior. Defined by mean (μ) and std (σ).
- **Uniform Distribution**: Used for initial random sampling. All values equally likely.
- **Exponential Distribution**: Used to model time between events or decay rates.
- **Binomial Distribution**: Used for binary outcomes (success/failure).

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='darkgrid', palette='muted', font_scale=1.2)
np.random.seed(42)

# Sample 1000 points from each distribution
normal_data     = np.random.normal(loc=0, scale=1, size=1000)
uniform_data    = np.random.uniform(low=0, high=1, size=1000)
exponential_data = np.random.exponential(scale=1.0, size=1000)
binomial_data   = np.random.binomial(n=10, p=0.5, size=1000)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Probability Distributions in Generative AI', fontsize=16, fontweight='bold')

# Normal Distribution
sns.histplot(normal_data, kde=True, ax=axes[0,0], color='steelblue')
axes[0,0].set_title('Normal Distribution (μ=0, σ=1)')
axes[0,0].set_xlabel('Value')
axes[0,0].axvline(np.mean(normal_data), color='red', linestyle='--', label=f'Mean={np.mean(normal_data):.2f}')
axes[0,0].legend()

# Uniform Distribution
sns.histplot(uniform_data, kde=True, ax=axes[0,1], color='orange')
axes[0,1].set_title('Uniform Distribution [0, 1]')
axes[0,1].set_xlabel('Value')

# Exponential Distribution
sns.histplot(exponential_data, kde=True, ax=axes[1,0], color='green')
axes[1,0].set_title('Exponential Distribution (scale=1)')
axes[1,0].set_xlabel('Value')

# Binomial Distribution
sns.histplot(binomial_data, kde=False, ax=axes[1,1], color='purple', discrete=True)
axes[1,1].set_title('Binomial Distribution (n=10, p=0.5)')
axes[1,1].set_xlabel('Number of Successes')

plt.tight_layout()
plt.savefig('exp1_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
print("Normal  -> Mean:", round(np.mean(normal_data),4), " Std:", round(np.std(normal_data),4))
print("Uniform -> Mean:", round(np.mean(uniform_data),4), " Std:", round(np.std(uniform_data),4))
```

## Result
- The **Normal distribution** is symmetric and bell-shaped, centered at mean=0.
- The **Uniform distribution** is flat — equal probability for all values.
- The **Exponential distribution** is right-skewed, modeling rare events.
- The **Binomial distribution** is discrete and approximately normal for large n.

## Conclusion
Different distributions serve different purposes in generative models. Normal distributions are most commonly used as priors in VAEs and diffusion models due to their mathematical tractability.
