```python
import numpy as np
import matplotlib.pyplot as plt

epochs=100

gen_loss=np.linspace(1.5,0.5,epochs)+np.random.normal(0,0.05,epochs)
disc_loss=np.linspace(0.5,1.0,epochs)+np.random.normal(0,0.05,epochs)

plt.plot(gen_loss,label="Generator")
plt.plot(disc_loss,label="Discriminator")
plt.legend()
plt.show()
```
