```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

true_mean = 5
true_std = 2

data = np.random.normal(true_mean, true_std, 1000)

estimated_mean = np.mean(data)
estimated_std = np.std(data)

print(estimated_mean, estimated_std)

plt.figure()
sns.histplot(data, kde=True, color="green")
plt.axvline(estimated_mean, color='red')
plt.axvline(true_mean, color='blue')
plt.show()
```
